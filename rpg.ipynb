```python
def vypocet_prumeru(znamky):
    if len(znamky) == 0:
        return 0
    else:
        return sum(znamky) / len(znamky)
```

```python
znamky_z_matematiky = []
znamky_z_cestiny = []
znamky_z_fyziky = []

while True:
    predmet = input("Zadej predmet (matematika = m, čeština = c, fyzika = f, konec = k ")
    if predmet == "m":
        znamka = float(input("Zadej známku z matematiky: "))
        if 1 <= znamka <= 5:
            print("Známka z matematiky je v pořádku.")
            znamky_z_matematiky.append(znamka)
        else:
            print("Známka z matematiky není v pořádku.")
    elif predmet == "c":
        znamka = float(input("Zadej známku z češtiny: "))
        if 1 <= znamka <= 5:
            print("Známka z češtiny je v pořádku.")
            znamky_z_cestiny.append(znamka)
        else:
            print("Známka z češtiny není v pořádku.")
    elif predmet == "f":
        znamka = float(input("Zadej známku z fyziky: "))
        if 1 <= znamka <= 5:
            print("Známka z fyziky je v pořádku.")
            znamky_z_fyziky.append(znamka)
        else:
            print("Známka z fyziky není v pořádku.")
    elif predmet == "k":
        break
    else:
        print("Neznámý predmet.")

if len(znamky_z_matematiky) > 0:
    prumer_m = sum(znamky_z_matematiky) / len(znamky_z_matematiky)
    print("Průměr z matematiky:", prumer_m)
if len(znamky_z_cestiny) > 0:
    prumer_c = sum(znamky_z_cestiny) / len(znamky_z_cestiny)
    print("Průměr z češtiny:", prumer_c)
if len(znamky_z_fyziky) > 0:
    prumer_f = sum(znamky_z_fyziky) / len(znamky_z_fyziky)
    print("Průměr z fyziky:", prumer_f)

vsechny = znamky_z_matematiky + znamky_z_cestiny + znamky_z_fyziky
if len(vsechny) > 0:
    print("Celkový průměr:", sum(vsechny) / len(vsechny))
```

```python
# 1. Inicializace proměnných na začátku programu
pokoj = [0]
inventar = []
penezenka = 0


# Pomocná funkce pro zjednodušení pokoje 0 a 1
def zeptej_se(text, platne):
    moznosti = [str(c) for c in platne]
    while True:
        volba = input(text + "Zadej volbu: ").strip()
        if volba in moznosti:
            return int(volba)
        print("Neplatná volba! Musíš zadat " + ", ".join(moznosti) + ".\n")


# Funkce POUZE pro Pokoj 0 a Pokoj 1
def pokoj_0():
    return zeptej_se(
        "\nProbudil ses ve svém pokoji.\n"
        "Jít do obýváku = 3\n"
        "Jít do skříně s oblečením = 2\n"
        "Jít zas spát = 0\n",
        [0, 2, 3],
    )


def pokoj_1():
    return zeptej_se(
        "\nVrátil ses do svého pokoje.\n"
        "Jít do obýváku = 3\n"
        "Jít do skříně s oblečením = 2\n"
        "Jít zas spát = 0\n",
        [0, 2, 3],
    )


# Zadání PINu (ukládá se jako text pro zachování počátečních nul)
while True:
    cislo = input("Zadej pin: ").strip()
    if cislo.isdigit():
        break
    print("To není číslo! Zadávat můžeš pouze číslice. Zkus to znovu.")

print(f"Skvělé, zadal jsi pin, zapamatuj si ho: {cislo}")

# Start hry
while True:
    start = input(
        "Hru budeš hrát tím, že budeš psát čísla do inputu. Abys začal/a hru,"
        " napiš 0: "
    )
    if start.strip() == "0":
        break
    print("Nezadal jsi 0, zkus to znovu.")

# 2. Hlavní herní smyčka
while True:
    aktualni_pokoj = pokoj[0]

    # Pokoj 0: Volá funkci pokoj_0
    if aktualni_pokoj == 0:
        pokoj[0] = pokoj_0()

    # Pokoj 1: Volá funkci pokoj_1
    elif aktualni_pokoj == 1:
        pokoj[0] = pokoj_1()

    # Pokoj 2: Skříň s oblečením
    elif aktualni_pokoj == 2:
        if "p" not in inventar:
            while True:
                volba = (
                    input("\nChceš se převléknout z pyžama? (ano / ne): ")
                    .strip()
                    .lower()
                )
                if volba in ["ano", "a"]:
                    inventar.append("p")
                    print("-> Oblečení 'p' bylo přidáno do inventáře.")
                    break
                elif volba in ["ne", "n"]:
                    print("-> Oblečení 'p' jsi nepřidal/a.")
                    break
                else:
                    print("Neplatná volba! Napiš 'ano' nebo 'ne'.")
        else:
            print("\n-> Oblečení 'p' už máš v inventáři.")

        print(f"Aktuální inventář: {inventar}")

        while True:
            try:
                komora = int(
                    input("\nVylézt zpět do pokoje = 1\nZadej volbu: ")
                )
                if komora == 1:
                    pokoj[0] = komora
                    break
                else:
                    print("Neplatná volba! Musíš zadat 1.\n")
            except ValueError:
                print("To není číslo! Zadávej pouze číslo 1.\n")

    # Pokoj 3: Obývák
    elif aktualni_pokoj == 3:
        while True:
            try:
                komora = int(
                    input(
                        "\nJsi v obývacím pokoji.\n"
                        "Podívat se na hodiny = 4\n"
                        "Udělat si snídani = 5\n"
                        "Jít zas do ložnice = 1\n"
                        "Jít z bytu = 6\n"
                        "Podívat se do btníku = 7\n"
                        "Zadej volbu: "
                    )
                )
                if komora in [1, 4, 5, 6, 7]:
                    pokoj[0] = komora
                    break
                else:
                    print("Neplatná volba! Musíš zadat 1, 4, 5, 6 nebo 7.\n")
            except ValueError:
                print("To není číslo! Zadávej pouze čísla 1, 4, 5, 6 nebo 7.\n")

    # Pokoj 4: Hodiny / Zaspal jsi
    elif aktualni_pokoj == 4:
        while True:
            try:
                komora = int(
                    input(
                        "\nZjistil jsi, že jsi zaspal a jdeš pozdě do práce.\n"
                        "Ignorovat a pokračovat se dnem = 3\n"
                        "Frčet do práce = 6\n"
                        "Zadej volbu: "
                    )
                )
                if komora in [3, 6]:
                    pokoj[0] = komora
                    break
                else:
                    print("Neplatná volba! Musíš zadat 3 nebo 6.\n")
            except ValueError:
                print("To není číslo! Zadávej pouze čísla 3 nebo 6.\n")

    # Pokoj 5: Snídaně
    elif aktualni_pokoj == 5:
        while True:
            try:
                komora = int(
                    input(
                        "\nZjistil jsi, že došlo všechno jídlo.\n"
                        "Ignorovat a pokračovat se dnem = 3\n"
                        "Frčet na nákup do Penny = 6\n"
                        "Zadej volbu: "
                    )
                )
                if komora in [3, 6]:
                    pokoj[0] = komora
                    break
                else:
                    print("Neplatná volba! Musíš zadat 3 nebo 6.\n")
            except ValueError:
                print("To není číslo! Zadávej pouze čísla 3 nebo 6.\n")

    # Pokoj 6: Odchod z bytu do předsíně
    elif aktualni_pokoj == 6:
        while True:
            try:
                komora = int(
                    input(
                        "\nVyšel jsi z bytu do předsíně, kde na tebe pokřikuje"
                        " zdejší bezdomovec.\n"
                        "Ignorovat a pokračovat ven = 10\n"
                        "Pustit se do hovoru s bezďákem = 9\n"
                        "Zdrhnout před bezdomovcem a schovat se v bytě = 3\n"
                        "Zadej volbu: "
                    )
                )
                if komora in [3, 9, 10]:
                    pokoj[0] = komora
                    break
                else:
                    print("Neplatná volba! Musíš zadat 3, 9 nebo 10.\n")
            except ValueError:
                print("To není číslo! Zadávej pouze čísla 3, 9 nebo 10.\n")

    # Pokoj 7: Botník
    elif aktualni_pokoj == 7:
        if "b" not in inventar:
            while True:
                volba = (
                    input(
                        "\nNašel jsi tam batoh, chceš si ho nandat a nosit po"
                        " bytě? (ano / ne): "
                    )
                    .strip()
                    .lower()
                )
                if volba in ["ano", "a"]:
                    inventar.append("b")
                    print("-> Batoh 'b' byl přidán do inventáře.")
                    break
                elif volba in ["ne", "n"]:
                    print("-> Batoh 'b' jsi nepřidal/a.")
                    break
                else:
                    print("Neplatná volba! Napiš 'ano' nebo 'ne'.")
        else:
            print("\n-> Batoh 'b' už máš v inventáři.")

        print(f"Aktuální inventář: {inventar}")

        while True:
            try:
                komora = int(
                    input("\nVylézt zpět do obýváku = 3\nZadej volbu: ")
                )
                if komora == 3:
                    pokoj[0] = komora
                    break
                else:
                    print("Neplatná volba! Musíš zadat 3.\n")
            except ValueError:
                print("To není číslo! Zadávej pouze číslo 3.\n")

    # Pokoj 8: Předsíň zvenku
    elif aktualni_pokoj == 8:
        while True:
            try:
                komora = int(
                    input(
                        "\nVlezl jsi do předsíně, kde na tebe pokřikuje zdejší"
                        " bezdomovec.\n"
                        "Ignorovat a pokračovat dovnitř = 3\n"
                        "Pustit se do hovoru s bezďákem = 9\n"
                        "Zdrhnout před bezdomovcem a schovat se venku před domem"
                        " = 10\n"
                        "Zadej volbu: "
                    )
                )
                if komora in [3, 9, 10]:
                    pokoj[0] = komora
                    break
                else:
                    print("Neplatná volba! Musíš zadat 3, 9 nebo 10.\n")
            except ValueError:
                print("To není číslo! Zadávej pouze čísla 3, 9 nebo 10.\n")

    # Pokoj 9: Bezdomovec
    elif aktualni_pokoj == 9:
        if "m" in inventar:
            while True:
                volba = (
                    input(
                        "\nPustil ses s bezdomovcem do hovoru a on chce drobné."
                        " Máš je u sebe. Dáš mu je? (ano / ne): "
                    )
                    .strip()
                    .lower()
                )
                if volba in ["ano", "a", "jo"]:
                    while True:
                        try:
                            kolik_drobnych = int(
                                input(
                                    "\nKolik peněz chceš dát bezdomovcovi? (máš"
                                    f" {penezenka} Kč): "
                                )
                            )
                            if 0 <= kolik_drobnych <= penezenka:
                                penezenka -= kolik_drobnych
                                print(
                                    f"-> Dal/a jsi bezďákovi {kolik_drobnych}"
                                    f" Kč. Tobě zbylo {penezenka} Kč."
                                )
                                break
                            else:
                                print(
                                    f"Můžeš dát pouze 0 až {penezenka} Kč!"
                                )
                        except ValueError:
                            print("To není číslo! Zadávej pouze číslo.")

                    if kolik_drobnych > 0:
                        print("-> Bezdomovec ti poděkoval a nechal tě jít.")
                        pokoj[0] = 10
                        break
                    else:
                        print("-> Nedal jsi bezďákovi ani korunu...")

                if volba in ["ne", "n", "no"] or (
                    volba in ["ano", "a", "jo"] and kolik_drobnych == 0
                ):
                    input(
                        "\nProtože jsi bezďákovi nedal drobné, ukončil tě. A je"
                        " konec hry... (Stiskni Enter)"
                    )
                    input("No... to neznělo moc přesvědčivě.")
                    input(
                        "Fakt, protože nezníš, že bys to pochopil, zníš"
                        " naštvaně."
                    )
                    input(
                        "Víš co? Já se slituji, protože jsem milý program."
                    )

                    while True:
                        try:
                            komora = int(
                                input(
                                    "Ano, ano, chápu tvé nadšení, jen napiš 0 a"
                                    " budeme předstírat, že to byl jen špatný"
                                    " sen: "
                                )
                            )
                            if komora == 0:
                                pokoj[0] = komora
                                break
                            else:
                                print("Já se ti snažím pomoct, napiš 0!\n")
                        except ValueError:
                            print("Já se ti snažím pomoct, napiš 0!\n")
                    break
                else:
                    print("Neplatná volba! Napiš 'ano' nebo 'ne'.")
        else:
            while True:
                try:
                    komora = int(
                        input(
                            "\nPustil ses s bezdomovcem do hovoru, on chce"
                            " drobné a ty sis uvědomil, že nemáš peníze.\n"
                            "Jít hledat peníze domů = 3\n"
                            "Jít ven = 10\n"
                            "Zadej volbu: "
                        )
                    )
                    if komora in [3, 10]:
                        pokoj[0] = komora
                        break
                    else:
                        print("Neplatná volba! Musíš zadat 3 nebo 10.\n")
                except ValueError:
                    print("To není číslo! Zadávej pouze čísla 3 nebo 10.\n")

    # Pokoj 10: Venku
    elif aktualni_pokoj == 10:
        while True:
            try:
                komora = int(
                    input(
                        "\nJsi venku, kam jdeš?\n"
                        "Do práce = 12\n"
                        "Vrátit se do domu = 8\n"
                        "Jít na nákup do Penny = 14\n"
                        "Jít si vybrat peníze z bankomatu = 11\n"
                        "Zadej volbu: "
                    )
                )
                if komora in [8, 11, 12, 14]:
                    pokoj[0] = komora
                    break
                else:
                    print("Neplatná volba! Musíš zadat 8, 11, 12 nebo 14.\n")
            except ValueError:
                print("To není číslo! Zadávej pouze čísla 8, 11, 12 nebo 14.\n")

    # Pokoj 11: Bankomat
    elif aktualni_pokoj == 11:
        if "m" not in inventar:
            while True:
                volba = (
                    input("\nChceš si vybrat peníze z bankomatu? (ano / ne): ")
                    .strip()
                    .lower()
                )
                if volba in ["ano", "a"]:
                    zadany_pin = input(
                        "Zadej PIN pro výběr z bankomatu: "
                    ).strip()
                    if zadany_pin == cislo:
                        inventar.append("m")
                        penezenka = 1000
                        print(
                            "-> Zadal jsi správný PIN! Peníze 'm' (1000 Kč) byly"
                            " přidány do inventáře."
                        )
                        break
                    else:
                        print("-> Špatný PIN! Peníze nebyly vybrány.")
                elif volba in ["ne", "n"]:
                    print("-> Peníze jsi nevybral, 'm' jsi nepřidal/a.")
                    break
                else:
                    print("Neplatná volba! Napiš 'ano' nebo 'ne'.")
        else:
            print("\n-> Peníze 'm' už máš v inventáři.")

        print(f"Aktuální inventář: {inventar}")

        while True:
            try:
                komora = int(
                    input("\nOdejít od bankomatu ven = 10\nZadej volbu: ")
                )
                if komora == 10:
                    pokoj[0] = komora
                    break
                else:
                    print("Neplatná volba! Musíš zadat 10.\n")
            except ValueError:
                print("To není číslo! Zadávej pouze číslo 10.\n")

    # Pokoj 12: Před prací
    elif aktualni_pokoj == 12:
        while True:
            try:
                komora = int(
                    input(
                        "\nJsi před prací, ale máš blbý pocit a nechce se ti"
                        " tam jít.\n"
                        "Jít do práce = 13\n"
                        "Vrátit se ven = 10\n"
                        "Zadej volbu: "
                    )
                )
                if komora in [10, 13]:
                    pokoj[0] = komora
                    break
                else:
                    print("Neplatná volba! Musíš zadat 10 nebo 13.\n")
            except ValueError:
                print("To není číslo! Zadávej pouze čísla 10 nebo 13.\n")

    # Pokoj 13: Práce
    elif aktualni_pokoj == 13:
        input("\nProtože jdeš pozdě, připrav si výmluvu pro šéfa... (Enter)")
        input("Šéf: 'Co tu děláš? Dostal jsi padáka před dvěma měsíci!' (Enter)")
        input(
            "Šéf: 'Ne, práci ti nevrátím, dostal jsi pětiměsíční plat.' (Enter)"
        )
        input("Šéf: 'Jdi domů!' (Enter)")
        input("Šéf: 'A nech těch protestů, mám toho plné zuby!!!!' (Enter)")

        while True:
            try:
                komora = int(input("\nJít domů/ven = 10\nZadej volbu: "))
                if komora == 10:
                    pokoj[0] = komora
                    break
                else:
                    print("Neplatná volba! Musíš zadat 10.\n")
            except ValueError:
                print("To není číslo! Zadávej pouze číslo 10.\n")

    # Pokoj 14: Penny
    elif aktualni_pokoj == 14:
        if "m" in inventar:
            while True:
                try:
                    komora = int(
                        input(
                            "\nChtěl jsi jít nakoupit do Penny, ale je plné"
                            " důchodců, kvalita jídla je na houby a Lidl je"
                            " obecně lepší.\n"
                            "Jít do Lidlu = 15\n"
                            "Jít domů = 10\n"
                            "Zadej volbu: "
                        )
                    )
                    if komora in [10, 15]:
                        pokoj[0] = komora
                        break
                    else:
                        print("Neplatná volba! Musíš zadat 10 nebo 15.\n")
                except ValueError:
                    print("To není číslo! Zadávej pouze čísla 10 nebo 15.\n")
        else:
            while True:
                try:
                    komora = int(
                        input(
                            "\nNemáš peníze, musíš si je vybrat nebo jít domů.\n"
                            "Jít ven = 10\n"
                            "Zadej volbu: "
                        )
                    )
                    if komora == 10:
                        pokoj[0] = komora
                        break
                    else:
                        print("Neplatná volba! Musíš zadat 10.\n")
                except ValueError:
                    print("To není číslo! Zadávej pouze číslo 10.\n")

    # Pokoj 15: Lidl
    elif aktualni_pokoj == 15:
        if "p" in inventar:
            while True:
                try:
                    komora = int(
                        input(
                            "\nVešel jsi do Lidlu, nakoupil jsi a uvědomil si:"
                            " 'Všude dobré ceny, v Lidlu nejlepší ceny!'\n"
                            "Ukončit hru = 16\n"
                            "Jít domů = 10\n"
                            "Zadej volbu: "
                        )
                    )
                    if komora in [10, 16]:
                        pokoj[0] = komora
                        break
                    else:
                        print("Neplatná volba! Musíš zadat 10 nebo 16.\n")
                except ValueError:
                    print("To není číslo! Zadávej pouze čísla 10 nebo 16.\n")
        else:
            while True:
                try:
                    komora = int(
                        input(
                            "\nCo šílíš? Jsi v pyžamu! Běž domů se převléknout,"
                            " Lidl má standardy!\n"
                            "Jít domů = 10\n"
                            "Zadej volbu: "
                        )
                    )
                    if komora == 10:
                        pokoj[0] = komora
                        break
                    else:
                        print("Neplatná volba! Musíš zadat 10.\n")
                except ValueError:
                    print("To není číslo! Zadávej pouze číslo 10.\n")

    # Pokoj 16: Konec hry
    elif aktualni_pokoj == 16:
        print("\nKonec hry. Gratuluji k dokončení!")
        break
```
